```python
import os, time, random, math
from math import isqrt, gcd
from Crypto.Util.number import getPrime, inverse
import matplotlib.pyplot as plt
import sympy

random.seed(123)
print("Ambiente pronto.")
```

## 1a. Fattorizzazione di Fermat — quando $p$ e $q$ sono troppo vicini

Se $p \approx q$, allora $n = pq$ è vicino a un quadrato perfetto: $n \approx a^2$ con
$a = \lceil\sqrt n\rceil$. Fermat cerca $a$ tale che $a^2 - n = b^2$ sia un quadrato perfetto:
allora $p = a-b$, $q = a+b$. Se $|p-q|$ è piccolo, questo converge in pochissime iterazioni.


```python
def gen_close_primes(bits=256, gap_search_limit=100000):
    p = getPrime(bits)
    q = p + 2
    tries = 0
    while not sympy.isprime(q) and tries < gap_search_limit:
        q += 2
        tries += 1
    return p, q

def fermat_factor(n, max_iter=200000):
    a = isqrt(n)
    if a * a < n:
        a += 1
    for _ in range(max_iter):
        b2 = a * a - n
        b = isqrt(b2)
        if b * b == b2:
            return a - b, a + b
        a += 1
    return None, None

p, q = gen_close_primes(256)
n = p * q
print(f"n = p*q, con p,q vicini (|p-q| = {abs(p-q)}), n ha {n.bit_length()} bit")

t0 = time.time()
f1, f2 = fermat_factor(n)
t_fermat = time.time() - t0
print(f"Fattorizzato in {t_fermat*1000:.1f} ms -> corretto: {sorted([f1,f2]) == sorted([p,q])}")
```

**Contromisura standard**: gli standard (es. FIPS 186) impongono che $|p-q|$ sia sufficientemente
grande (tipicamente richiedendo che $p$ e $q$ differiscano per più di $2^{bits/2 - 100}$ circa) — una
condizione oggi enforced automaticamente dai generatori di chiavi RSA seri.


## 1b. Common modulus attack — non condividere mai lo stesso $n$

Se due utenti condividono lo stesso modulo $n$ ma esponenti pubblici diversi e **coprimi** $e_1, e_2$, e lo
stesso messaggio $m$ viene cifrato per entrambi, chiunque intercetti $c_1, c_2$ può recuperare $m$ **senza
fattorizzare $n$**, usando l'identità di Bézout: $a e_1 + b e_2 = 1 \Rightarrow c_1^a c_2^b \equiv m \pmod n$.


```python
def extended_gcd(a, b):
    if a == 0:
        return b, 0, 1
    g, x1, y1 = extended_gcd(b % a, a)
    return g, y1 - (b // a) * x1, x1

def common_modulus_attack(n, e1, c1, e2, c2):
    g, a, b = extended_gcd(e1, e2)
    assert g == 1, "e1, e2 devono essere coprimi"
    if a < 0:
        c1 = inverse(c1, n); a = -a
    if b < 0:
        c2 = inverse(c2, n); b = -b
    return (pow(c1, a, n) * pow(c2, b, n)) % n

p2, q2 = getPrime(256), getPrime(256)
n2 = p2 * q2
phi2 = (p2 - 1) * (q2 - 1)
e1 = 17
while gcd(e1, phi2) != 1: e1 += 2
e2 = 65537
while gcd(e2, phi2) != 1: e2 += 2

secret_message = 987654321
c1 = pow(secret_message, e1, n2)
c2 = pow(secret_message, e2, n2)

recovered = common_modulus_attack(n2, e1, c1, e2, c2)
print("Eve intercetta solo (n, e1, c1) e (n, e2, c2), MAI la fattorizzazione di n.")
print("Messaggio recuperato:", recovered, "| corretto:", recovered == secret_message)
```

**Contromisura standard**: ogni utente deve avere il proprio modulo $n$ generato indipendentemente. Il
modulo non va mai condiviso, nemmeno "per efficienza".


## 1c. Attacco di Wiener — l'esponente privato $d$ non può essere troppo piccolo

Se $d < \tfrac{1}{3} n^{1/4}$ (circa), l'attacco di Wiener recupera $d$ in tempo polinomiale usando le
**frazioni continue**: $e/n$ ha come uno dei suoi convergenti la frazione $k/d$.


```python
from sympy import continued_fraction_convergents, continued_fraction_iterator, Rational

def gen_wiener_vulnerable(bits=512):
    while True:
        p = getPrime(bits // 2)
        q = getPrime(bits // 2)
        if p == q:
            continue
        n = p * q
        phi = (p - 1) * (q - 1)
        d_bits = bits // 4 - 5   # esponente privato "troppo piccolo"
        d = random.getrandbits(d_bits) | 1
        try:
            e = inverse(d, phi)
        except Exception:
            continue
        if e > 1:
            return n, e, d

def wiener_attack(n, e):
    cf = list(continued_fraction_iterator(Rational(e, n)))
    for conv in continued_fraction_convergents(cf):
        k, d = conv.p, conv.q
        if k == 0:
            continue
        if (e * d - 1) % k != 0:
            continue
        phi = (e * d - 1) // k
        b = n - phi + 1
        disc = b * b - 4 * n
        if disc < 0:
            continue
        sq = isqrt(disc)
        if sq * sq != disc:
            continue
        x1, x2 = (b + sq) // 2, (b - sq) // 2
        if x1 * x2 == n:
            return d
    return None

n3, e3, d3 = gen_wiener_vulnerable(512)
print(f"n ha {n3.bit_length()} bit, d 'debole' ha {d3.bit_length()} bit (troppo piccolo)")

t0 = time.time()
d_found = wiener_attack(n3, e3)
t_wiener = time.time() - t0
print(f"d recuperato in {t_wiener*1000:.1f} ms via frazioni continue: {d_found == d3}")
```

**Contromisura standard**: gli standard impongono $d > n^{0.5}$ circa (in pratica: si sceglie prima $e$
piccolo e standard come $65537$, e si lascia che $d$ venga fuori "grande" per costruzione).


## 1d. Broadcast attack di Håstad — $e$ piccolo senza padding

Se lo stesso messaggio $m$ viene cifrato con $e=3$ per **tre** destinatari con moduli diversi $n_1,n_2,n_3$
(coprimi tra loro) e **senza padding randomizzato**, il Teorema Cinese del Resto ricostruisce $m^3 \bmod
(n_1 n_2 n_3)$; se $m^3 < n_1 n_2 n_3$ (quasi sempre vero), basta la radice cubica intera.


```python
from sympy.ntheory.modular import crt

def hastad_broadcast(cs, ns):
    combined = crt(ns, cs)[0]
    lo, hi = 0, combined
    while lo < hi:      # ricerca binaria della radice cubica intera
        mid = (lo + hi) // 2
        if mid ** 3 < combined:
            lo = mid + 1
        else:
            hi = mid
    return lo

secret_broadcast = 20260922  # es. una data, un codice, un piccolo segreto broadcastato a 3 destinatari
ns_list, cs_list = [], []
for _ in range(3):
    pp, qq = getPrime(256), getPrime(256)
    nn = pp * qq
    ns_list.append(nn)
    cs_list.append(pow(secret_broadcast, 3, nn))   # e = 3, NESSUN padding

recovered_b = hastad_broadcast(cs_list, ns_list)
print("Messaggio recuperato via CRT + radice cubica, senza fattorizzare nulla:", recovered_b)
print("Corretto:", recovered_b == secret_broadcast)
```

**Contromisura standard**: **mai** cifrare senza padding randomizzato (OAEP, visto Giorno 1) — con OAEP
lo stesso messaggio produce ciphertext completamente diversi per ogni destinatario, e l'attacco CRT non si
applica più.

### Riepilogo della sezione

| Attacco | Difetto sfruttato | Contromisura diventata standard |
|---|---|---|
| Fermat | $p,q$ troppo vicini | vincolo minimo su $\vert p-q\vert$ |
| Common modulus | $n$ condiviso | un $n$ per utente, mai condiviso |
| Wiener | $d$ troppo piccolo | $d$ sufficientemente grande (e piccolo e fisso) |
| Håstad broadcast | $e$ piccolo, niente padding | padding randomizzato (OAEP) |

Nessuno di questi attacchi rompe la fattorizzazione: rompono **scelte di parametri**. Da qui il messaggio
del corso: "la primitiva non basta, serve un parameter set correttamente dimensionato e continuamente
rivalutato".


## 2. Perché il quantum computer cambia le regole

Fattorizzazione classica: il miglior algoritmo noto (GNFS) ha complessità **sub-esponenziale**
$L_n[1/3, c] = \exp\big((c+o(1))(\ln n)^{1/3}(\ln\ln n)^{2/3}\big)$.
L'algoritmo di Shor su un computer quantistico risolve fattorizzazione (e logaritmo discreto, quindi anche
DH/ECC) in tempo **polinomiale**. Visualizziamo la crescita relativa.


```python
def gnfs_cost_log10(bits):
    n_ln = bits * math.log(2)
    L = math.exp((1.923 + 0.01) * (n_ln ** (1/3)) * (math.log(n_ln) ** (2/3)))
    return math.log10(L)

def shor_cost_log10(bits):
    # costo polinomiale ~ bits^3 (stima approssimativa per il confronto visivo)
    return math.log10(bits ** 3)

bit_range = [512, 1024, 1536, 2048, 3072, 4096, 8192]
gnfs = [gnfs_cost_log10(b) for b in bit_range]
shor = [shor_cost_log10(b) for b in bit_range]

plt.figure(figsize=(6,4))
plt.plot(bit_range, gnfs, marker='o', label="Fattorizzazione classica (GNFS), sub-esponenziale")
plt.plot(bit_range, shor, marker='s', label="Algoritmo di Shor, polinomiale")
plt.xlabel("dimensione del modulo RSA (bit)")
plt.ylabel("log10(operazioni stimate)")
plt.title("Il gap che aumentare la dimensione della chiave non può colmare contro Shor")
plt.legend(); plt.grid(True, alpha=0.3); plt.show()
```

**Takeaway**: contro un attaccante classico, raddoppiare la dimensione della chiave RSA alza
enormemente il costo (curva sub-esponenziale ma comunque ripida). Contro Shor, la curva del costo cresce
solo **polinomialmente**: non esiste una dimensione di chiave "abbastanza grande" che risolva il problema
—serve cambiare **famiglia di problema matematico**, non solo i parametri. Questo è esattamente il salto
concettuale rispetto alla sezione 1: lì bastava correggere un parametro, qui serve una primitiva diversa.


## 3. Quanto dovrebbe crescere RSA per "resistere" a un quantum computer? (spoiler: non ha senso)

Anche solo per intuizione quantitativa, mostriamo le stime pubblicate (Gidney & Ekerå, 2019) delle risorse
quantistiche necessarie per fattorizzare RSA-2048 con Shor, confrontate con le dimensioni di chiave PQC
equivalenti a parità di *livello di sicurezza* NIST.


```python
# Valori di riferimento (letteratura pubblica, per la sola discussione in aula):
# Gidney & Ekera (2019): RSA-2048 rompibile con ~20 milioni di qubit fisici rumorosi
# in circa 8 ore, con l'hardware ipotizzato nel paper.
scenarios = [
    ("RSA-2048 (classico, GNFS)",        "~10^9 anni-CPU equivalenti (stima colloquiale)"),
    ("RSA-2048 (quantistico, Shor)",     "~20 milioni di qubit fisici, ~8 ore (Gidney-Ekera 2019)"),
    ("RSA-4096 (quantistico, Shor)",     "qubit necessari cresce polinomialmente: ~4x le operazioni, non 'sicuro'"),
]
for name, cost in scenarios:
    print(f"{name:<32} -> {cost}")

print()
print("Confronto dimensioni chiave/ciphertext a parità di NIST security level (categoria 1, ~AES-128):")
comparison = [
    ("RSA-3072 (classico)",        "chiave pubblica ~384 byte", "NON sicuro contro quantum"),
    ("ML-KEM-512 (Kyber, PQC)",    "chiave pubblica 800 byte, ciphertext 768 byte", "sicuro contro quantum (lattice)"),
]
for name, size, note in comparison:
    print(f"{name:<28} {size:<48} {note}")
```

**Takeaway della sezione**: aumentare la dimensione di RSA non è una risposta valida al quantum
computing (il vantaggio di Shor è strutturale, non quantitativo). Serve cambiare il problema matematico
sottostante — da qui i **reticoli**, i **codici correttori**, gli **hash-based scheme**: famiglie per cui
non è noto (a oggi) alcun algoritmo quantistico efficiente.


## 4. Un problema hard per i reticoli — LWE, visualizzato

*Learning With Errors* (LWE): dato un sistema lineare $A\mathbf{s} + \mathbf{e} \approx \mathbf{b}
\pmod q$ con un **piccolo rumore** $\mathbf{e}$ aggiunto apposta, recuperare $\mathbf{s}$ è difficile
(anche per un quantum computer) — mentre senza rumore sarebbe algebra lineare elementare, banale da
risolvere. Il rumore è la "trapdoor inversa": chi conosce una struttura segreta (una base "buona" del
reticolo) può filtrarlo, chi non la conosce no.


```python
def toy_lwe_demo(n=6, q=97, num_samples=10, noise_bound=2):
    s = [random.randrange(q) for _ in range(n)]           # segreto
    samples = []
    for _ in range(num_samples):
        a = [random.randrange(q) for _ in range(n)]
        e = random.randint(-noise_bound, noise_bound)      # piccolo rumore
        b = (sum(ai * si for ai, si in zip(a, s)) + e) % q
        samples.append((a, b))
    return s, samples

def solve_without_noise(samples, n, q):
    """Se NON ci fosse rumore, n campioni bastano: algebra lineare elementare (qui via forza bruta
    su un piccolo esempio, solo per mostrare quanto sarebbe banale SENZA il rumore)."""
    import itertools
    a0, b0 = samples[0]
    for candidate in itertools.product(range(q), repeat=n):
        if sum(ai*ci for ai,ci in zip(a0, candidate)) % q == b0:
            # verifica sugli altri campioni (nel caso senza rumore basterebbero n campioni indipendenti,
            # qui semplificato per la demo con n molto piccolo)
            ok = all(sum(ai*ci for ai,ci in zip(a,candidate)) % q == b for a,b in samples[1:3])
            if ok:
                return candidate
    return None

n_dim, q_mod = 3, 31   # piccolissimo, SOLO per rendere l'esempio "senza rumore" trattabile in aula
s_true, samples_noisy = toy_lwe_demo(n=n_dim, q=q_mod, num_samples=5, noise_bound=0)  # rumore=0 per questa parte
print("Segreto reale:", s_true)
t0 = time.time()
found = solve_without_noise(samples_noisy, n_dim, q_mod)
print(f"SENZA rumore: risolto per forza bruta in {time.time()-t0:.2f}s -> {found}")
```

```python
# Ora con rumore: la stessa forza bruta fallisce perché la relazione non è più esatta.
s_true2, samples_with_noise = toy_lwe_demo(n=n_dim, q=q_mod, num_samples=5, noise_bound=2)
found2 = solve_without_noise(samples_with_noise, n_dim, q_mod)
print("CON anche solo un piccolo rumore, la stessa ricerca esatta:", found2, "(fallisce: nessuna soluzione esatta)")
print()
print("Per risolvere LWE con rumore servono tecniche di riduzione di reticolo (es. LLL/BKZ) che, anche")
print("nella versione quantistica nota, restano costose: questo è il problema su cui si basano ML-KEM (Kyber),")
print("ML-DSA (Dilithium) e la maggior parte dei candidati NIST PQC.")
```

**Takeaway**: il rumore trasforma un problema di algebra lineare banale in un problema di ricerca in
uno spazio ad alta dimensione, per cui non sono noti algoritmi quantistici efficienti — a differenza di
fattorizzazione e logaritmo discreto, che Shor risolve in tempo polinomiale.


## 5. KEM giocattolo (stile ML-KEM/Kyber) vs RSA-OAEP

Un **KEM** (Key Encapsulation Mechanism) non cifra un messaggio a scelta: incapsula una chiave *casuale*
generata dal mittente. È il paradigma dominante in PQC. Costruiamo una versione LWE-based fortemente
semplificata (parametri giocattolo, NON sicura, solo per capire il meccanismo) e la confrontiamo con
RSA-OAEP del Giorno 1 su dimensioni e tempi.


```python
class ToyLWEKEM:
    """KEM LWE-based ridotto all'osso, per didattica. Non usare mai in produzione:
    dimensioni/parametri troppo piccoli per essere sicuri. Codifica del bit alla Regev:
    0 -> vicino a 0, 1 -> vicino a q/2, cosi' il rumore non fa sconfinare la decodifica."""
    def __init__(self, n=64, q=3329, noise_bound=2):
        self.n, self.q, self.noise_bound = n, q, noise_bound

    def keygen(self):
        s = [random.randrange(self.q) for _ in range(self.n)]     # chiave privata
        A = [[random.randrange(self.q) for _ in range(self.n)] for _ in range(self.n)]
        e = [random.randint(-self.noise_bound, self.noise_bound) for _ in range(self.n)]
        b = [(sum(A[i][j]*s[j] for j in range(self.n)) + e[i]) % self.q for i in range(self.n)]
        return (A, b), s     # (chiave pubblica, chiave privata)

    def encapsulate(self, pub, bit=None):
        A, b = pub
        if bit is None:
            bit = random.randint(0, 1)          # il bit incapsulato diventa la chiave condivisa
        r = [random.randint(0, 1) for _ in range(self.n)]          # vettore binario piccolo
        u = [sum(r[i]*A[i][j] for i in range(self.n)) % self.q for j in range(self.n)]
        v = (sum(r[i]*b[i] for i in range(self.n)) + bit*(self.q // 2)) % self.q
        return (u, v), bit

    def decapsulate(self, priv, ct):
        s = priv
        u, v = ct
        pred = (v - sum(u[j]*s[j] for j in range(self.n))) % self.q
        dist0, dist_half = min(pred, self.q - pred), abs(pred - self.q // 2)
        return 0 if dist0 < dist_half else 1    # decodifica per arrotondamento: vicino a 0 o a q/2?


kem = ToyLWEKEM(n=64, q=3329, noise_bound=2)   # n ridotto per far girare la demo in aula in tempo utile
t0 = time.time()
pub, priv = kem.keygen()
t_keygen = time.time() - t0

t0 = time.time()
ct, ss_sender = kem.encapsulate(pub)
t_encap = time.time() - t0

t0 = time.time()
ss_receiver = kem.decapsulate(priv, ct)
t_decap = time.time() - t0

print(f"Keygen: {t_keygen*1000:.2f} ms | Encaps: {t_encap*1000:.2f} ms | Decaps: {t_decap*1000:.2f} ms")
print(f"Bit condiviso mittente: {ss_sender} | destinatario: {ss_receiver} | coincidono: {ss_sender == ss_receiver}")

# Verifica di affidabilità su molte esecuzioni (con parametri ben dimensionati il fallimento è trascurabile,
# esattamente come per ML-KEM reale, che usa codici di correzione d'errore per azzerarlo)
ok = sum(kem.decapsulate(priv, (ct := kem.encapsulate(pub))[0]) == ct[1] for _ in range(200))
print(f"Affidabilità su 200 round: {ok}/200 decapsulazioni corrette")
```

*(Nella pratica, ML-KEM incapsula molti bit in parallelo tramite codici di correzione d'errore e
compressione, con $n\approx 256$-$1024$ e $q$ standardizzato — qui isoliamo il meccanismo concettuale con
un solo bit per chiarezza in aula.)*


```python
# Confronto concettuale finale con RSA-OAEP (dati indicativi standard, non ricalcolati qui)
comparison2 = [
    ("RSA-2048 + OAEP",     "256 byte chiave pubblica", "256 byte ciphertext", "vulnerabile a Shor"),
    ("ML-KEM-512 (Kyber)",  "800 byte chiave pubblica", "768 byte ciphertext", "resistente a Shor (LWE)"),
    ("ML-KEM-768 (Kyber)",  "1184 byte chiave pubblica","1088 byte ciphertext","livello ~AES-192"),
]
print(f"{'Schema':<20}{'Chiave pubblica':<22}{'Ciphertext':<20}{'Nota'}")
for row in comparison2:
    print(f"{row[0]:<20}{row[1]:<22}{row[2]:<20}{row[3]}")
```

**Takeaway finale del corso**: lo schema architetturale — *usa l'asimmetrica solo per instaurare un
segreto condiviso, poi passa al simmetrico per i dati* — è identico dal Giorno 1 (RSA-OAEP) al Giorno 3
(ML-KEM): cambia solo il problema matematico su cui poggia la trapdoor. Le chiavi/ciphertext PQC sono più
grandi di RSA/ECC a parità di livello di sicurezza classico, ed è il prezzo da pagare per la resistenza
quantistica: da qui l'interesse per gli **approcci ibridi** (es. X25519 + ML-KEM) discussi nella parte
teorica sulla migrazione e sulla crypto-agility.


## Open Quantum Safe

Spiegare openquantum safe e la sua facilita' di integrare primtive post quantum direttamente nei protocolli
