# 1. Il processo NIST PQC

Il processo di standardizzazione è iniziato con la call for proposals del 2016 e la submission dei candidati nel 2017.

Il percorso può essere sintetizzato in:

```text
2016  Call for proposals
  ↓
2017  82 submissions
  ↓
2019  Round 2 — 26 candidates
  ↓
2020  Round 3 — 7 finalists + 8 alternates
  ↓
2022  Selezione degli algoritmi per la standardizzazione
  ↓
2024  Pubblicazione dei primi 3 FIPS
  ↓
2025  HQC selezionato per futura standardizzazione
  ↓
2026  Falcon selezionato per futura standardizzazione
```

La timeline è volutamente sintetica: l'obiettivo è capire **come si è arrivati agli standard**, non ripercorrere ogni submission.


## Timeline visuale

````python
import matplotlib.pyplot as plt

events = [
    (2016, "PQC Call", ""),
    (2017, "Round 1", "82 submissions"),
    (2019, "Round 2", "26 candidates"),
    (2020, "Round 3", "7 finalists + 8 alternates"),
    (2022, "Selection", "Algorithms selected"),
    (2024, "Standardization", "ML-KEM · ML-DSA · SLH-DSA"),
    (2025, "HQC", "Selected for standardization"),
    (2026, "Falcon", "Selected for future FIPS"),
]

fig, ax = plt.subplots(figsize=(16, 5))

# Timeline
ax.hlines(
    0,
    events[0][0] - 0.1,
    events[-1][0] + 0.2,
    linewidth=2
)

for i, (year, stage, detail) in enumerate(events):

    # Punto sulla timeline
    ax.scatter(
        year,
        0,
        s=120,
        zorder=3
    )

    # Alterniamo sopra/sotto
    y = 0.18 if i % 2 == 0 else -0.18
    va = "bottom" if y > 0 else "top"

    # Data + round sulla stessa riga
    ax.text(
        year,
        y,
        f"{year}  |  {stage}",
        ha="center",
        va=va,
        fontsize=10,
        fontweight="bold"
    )

    # Dettaglio sotto la data/round
    detail_y = y + (0.11 if y > 0 else -0.11)

    ax.text(
        year,
        detail_y,
        detail,
        ha="center",
        va=va if y < 0 else "top",
        fontsize=9
    )

# Evidenziamo il 2024
ax.axvspan(
    2023.7,
    2024.3,
    alpha=0.08
)

ax.text(
    2024,
    0.43,
    "NIST PQC Standards",
    ha="center",
    fontsize=11,
    fontweight="bold"
)

ax.set_xlim(2015.5, 2026.5)
ax.set_ylim(-0.48, 0.48)

ax.set_xticks(range(2016, 2027))
ax.set_yticks([])

ax.set_title(
    "NIST Post-Quantum Cryptography Standardization Timeline",
    fontsize=20
)

ax.spines["left"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

plt.tight_layout()
plt.show()
````

# 2. Cosa ha prodotto il processo NIST?

## Standard pubblicati

### FIPS 203 — ML-KEM

**ML-KEM** è il KEM standardizzato da NIST.

Famiglia:

```text
Module-Lattice
      │
      └── ML-KEM
```

Parametri:

```text
ML-KEM-512
ML-KEM-768
ML-KEM-1024
```

API concettuale:

```text
KeyGen
  ↓
public key / secret key
  ↓
Encaps ──────────→ ciphertext + shared secret
                         ↓
                      Decaps
                         ↓
                   shared secret
```


# 3. Digital Signatures

## FIPS 204 — ML-DSA

Famiglia:

```text
Module-Lattice
      │
      └── ML-DSA
```

Varianti:

```text
ML-DSA-44
ML-DSA-65
ML-DSA-87
```

API:

```text
KeyGen
  ↓
public key / secret key

secret key + message
        ↓
      Sign
        ↓
   signature

public key + message + signature
        ↓
      Verify
        ↓
   valid / invalid
```

---

## FIPS 205 — SLH-DSA

SLH-DSA appartiene alla famiglia delle **hash-based signatures**.

```text
Hash-based
    │
    └── SLH-DSA
```

È particolarmente interessante nel contesto della diversificazione delle assunzioni crittografiche.


# 4. Algoritmi selezionati per la standardizzazione futura

Il processo non si è fermato ai primi tre FIPS.

## HQC

HQC è un KEM **code-based** selezionato da NIST nel 2025 per una futura standardizzazione.

```text
Code-based
    │
    └── HQC
```

## Falcon

Falcon è una signature basata su reticoli, selezionata da NIST nel 2024 per una futura standardizzazione, con successiva pianificazione del relativo FIPS.

```text
Lattice / NTRU
      │
      └── Falcon
```

Questi algoritmi sono importanti anche come esempio della strategia di NIST di mantenere **diversità nelle assunzioni matematiche**.

# 8. Verifichiamo l'installazione

Importiamo il binding Python.

````python
import oqs

print("liboqs-python import: OK")
````

## Quali KEM sono disponibili?

La lista dipende dalla versione e dalla configurazione della build di `liboqs`.

````python
kem_algorithms = oqs.get_enabled_kem_mechanisms()

print(f"Available KEMs: {len(kem_algorithms)}")
for algorithm in kem_algorithms:
    print(" -", algorithm)
````

## Quali signature sono disponibili?

````python
sig_algorithms = oqs.get_enabled_sig_mechanisms()

print(f"Available signature algorithms: {len(sig_algorithms)}")
for algorithm in sig_algorithms:
    print(" -", algorithm)
````

# 9. ML-KEM — prima prova

Partiamo dallo standard principale: **ML-KEM**.

````python
kem_name = "ML-KEM-768"

if kem_name not in kem_algorithms:
    raise RuntimeError(f"{kem_name} non è disponibile nella build corrente")

kem = oqs.KeyEncapsulation(kem_name)

print("Algorithm:", kem_name)
print("Details:")
print(kem.details)
````

## Key generation

````python
public_key = kem.generate_keypair()

print("Public key length:", len(public_key), "bytes")
````

## Encapsulation

````python
ciphertext, shared_secret = kem.encap_secret(public_key)

print("Ciphertext length:", len(ciphertext), "bytes")
print("Shared secret length:", len(shared_secret), "bytes")
print("Shared secret:", shared_secret.hex())
````

## Decapsulation

````python
shared_secret_2 = kem.decap_secret(ciphertext)

print("Recovered secret:", shared_secret_2.hex())
print("Match:", shared_secret == shared_secret_2)

assert shared_secret == shared_secret_2
````

# 12. ML-DSA — Digital Signature

Ora passiamo alle firme.

````python
sig_name = "ML-DSA-65"

if sig_name not in sig_algorithms:
    raise RuntimeError(f"{sig_name} non è disponibile nella build corrente")

sig = oqs.Signature(sig_name)

print("Algorithm:", sig_name)
print("Details:")
print(sig.details)
````

## Key generation

````python
public_key_sig = sig.generate_keypair()

print("Public key length:", len(public_key_sig), "bytes")
````

## Sign

````python
message = b"Post-Quantum Cryptography with Open Quantum Safe"

signature = sig.sign(message)

print("Message length:", len(message))
print("Signature length:", len(signature), "bytes")
````

## Verify

````python
valid = sig.verify(
    message,
    signature,
    public_key_sig
)

print("Signature valid:", valid)
assert valid
````

## Modifichiamo il messaggio

````python
tampered_message = message + b" modified"

valid = sig.verify(
    tampered_message,
    signature,
    public_key_sig
)

print("Signature valid after modification:", valid)
assert not valid
````

# 13. Dimensioni dei parametri

Una delle differenze più evidenti rispetto a RSA è la dimensione degli oggetti crittografici.

Possiamo raccogliere i parametri direttamente dalla libreria.

````python
def kem_summary(name):
    kem = oqs.KeyEncapsulation(name)
    details = kem.details
    return {
        "Algorithm": name,
        "Public key": details.get("length_public_key"),
        "Secret key": details.get("length_secret_key"),
        "Ciphertext": details.get("length_ciphertext"),
        "Shared secret": details.get("length_shared_secret"),
    }


for name in ["ML-KEM-512", "ML-KEM-768", "ML-KEM-1024"]:
    if name in kem_algorithms:
        print(kem_summary(name))
````

# 14. Benchmark: KeyGen / Encaps / Decaps

Ora passiamo dalla correttezza alla performance.

```text
Algorithm
    │
    ├── KeyGen
    ├── Encaps
    └── Decaps
```

Il benchmark seguente è volutamente semplice: serve a vedere l'ordine di grandezza e confrontare le varianti.


````python
import time
import statistics


def benchmark_kem(name, iterations=100):
    timings = {
        "keygen": [],
        "encaps": [],
        "decaps": [],
    }

    for _ in range(iterations):
        kem = oqs.KeyEncapsulation(name)

        t0 = time.perf_counter()
        public_key = kem.generate_keypair()
        t1 = time.perf_counter()

        ciphertext, shared_secret = kem.encap_secret(public_key)
        t2 = time.perf_counter()

        recovered = kem.decap_secret(ciphertext)
        t3 = time.perf_counter()

        assert shared_secret == recovered

        timings["keygen"].append(t1 - t0)
        timings["encaps"].append(t2 - t1)
        timings["decaps"].append(t3 - t2)

    return {
        operation: statistics.median(values) * 1000
        for operation, values in timings.items()
    }


for name in ["ML-KEM-512", "ML-KEM-768", "ML-KEM-1024"]:
    if name in kem_algorithms:
        print(name, benchmark_kem(name))
````

# 15. Benchmark delle firme

Per le signature misuriamo:

```text
KeyGen
Sign
Verify
```

````python
def benchmark_signature(name, iterations=100):
    timings = {
        "keygen": [],
        "sign": [],
        "verify": [],
    }

    message = b"benchmark message"

    for _ in range(iterations):
        sig = oqs.Signature(name)

        t0 = time.perf_counter()
        public_key = sig.generate_keypair()
        t1 = time.perf_counter()

        signature = sig.sign(message)
        t2 = time.perf_counter()

        valid = sig.verify(message, signature, public_key)
        t3 = time.perf_counter()

        assert valid

        timings["keygen"].append(t1 - t0)
        timings["sign"].append(t2 - t1)
        timings["verify"].append(t3 - t2)

    return {
        operation: statistics.median(values) * 1000
        for operation, values in timings.items()
    }


for name in ["ML-DSA-44", "ML-DSA-65", "ML-DSA-87"]:
    if name in sig_algorithms:
        print(name, benchmark_signature(name))
````
