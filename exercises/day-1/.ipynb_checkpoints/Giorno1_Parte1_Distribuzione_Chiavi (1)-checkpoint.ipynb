# Giorno 1 — Parte 1: Il problema della distribuzione delle chiavi

Primo file dell'esercitazione del Giorno 1. Percorso:

1. L'assunto nascosto della crittografia simmetrica: le chiavi sono **già condivise**.
2. La soluzione ingenua (una chiave per ogni coppia di peer) e la sua **esplosione combinatoria**,
   esplorata con un grafo interattivo.
3. Si può usare la *sola* crittografia simmetrica per generare un segreto condiviso su un canale
   insicuro? Sì (**Merkle Puzzles**), ma con prestazioni pessime.

> Alla fine di questo file resta aperta una domanda che il prossimo file (fondamenti RSA / trapdoor)
> risolverà: serve un meccanismo con un gap di sicurezza *migliore* di quadratico tra chi possiede il
> segreto e chi lo attacca.


````
import os, time, random, itertools
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, IntSlider

random.seed(42)
print("Ambiente pronto.")
````

## 1. L'assunto nascosto: le chiavi sono già condivise

Uno schema simmetrico (AES, ChaCha20, ...) presuppone che Alice e Bob possiedano **già** la stessa chiave
segreta $K$ prima ancora di iniziare a comunicare. La cifratura/decifratura è il problema che questi
schemi risolvono — ma **come fanno Alice e Bob ad avere $K$ in comune**, se non si sono mai incontrati e
comunicano solo su un canale che un attaccante può origliare? Questo è il **problema della distribuzione
delle chiavi**, ed è il filo conduttore di tutto il corso.


### Schema a blocchi: come funziona (concettualmente) la crittografia simmetrica

```
                         canale INSICURO (Eve puo' leggere tutto qui)
                    ┌───────────────────────────────────────────────┐
                    │                                                │
   Alice            │                  ciphertext C                 │            Bob
┌──────────┐   ┌────┴─────┐                                   ┌─────┴────┐   ┌──────────┐
│ messaggio│──▶│  Encrypt │──────────────────────────────────▶│  Decrypt │──▶│ messaggio│
│    M     │   │  E(K, M) │                                   │  D(K, C) │   │    M     │
└──────────┘   └────┬─────┘                                   └─────┬────┘   └──────────┘
                     │                                               │
                     └───────────────────  K  ──────────────────────┘
                          stessa chiave segreta, GIA' condivisa
                          da qualche canale/momento precedente
                          (mai transitata sul canale insicuro!)

                                     ▲
                                     │
                             Eve vede solo C:
                          senza K, non ricava M
                     ma... DA DOVE VIENE K per Alice e Bob?
                          <- e' esattamente la domanda
                             che questo file affronta
```

Il punto chiave dello schema: `K` **non passa mai** sul canale insicuro — deve essere già nota a
entrambi *prima* che lo scambio di `C` abbia luogo. Tutta la sicurezza del sistema poggia su
un'ipotesi che lo schema stesso non spiega come realizzare.


````
# Cifratura simmetrica: banale SE la chiave e' gia' condivisa
shared_key = os.urandom(32)  # 256 bit, gia' nota sia ad Alice sia a Bob

message = b"Attacco alle 6"
cipher = AES.new(shared_key, AES.MODE_GCM)
nonce = cipher.nonce
ct, tag = cipher.encrypt_and_digest(message)

# Bob, che possiede la STESSA chiave, decifra senza problemi
decipher = AES.new(shared_key, AES.MODE_GCM, nonce=nonce)
recovered = decipher.decrypt_and_verify(ct, tag)
print("Messaggio decifrato da Bob:", recovered)
print()
print("Tutto facile... ma DA DOVE VIENE 'shared_key'? Questa domanda e' il vero problema.")
````

## 2. La soluzione ingenua: una chiave per ogni coppia di peer

L'idea più semplice: ogni coppia di persone che potrebbe voler comunicare si scambia una chiave *una
tantum*, magari di persona, in anticipo (es. su una chiavetta USB, a voce). Con $N$ peer nella rete, quante
chiavi servono in totale, se ognuno vuole poter parlare con chiunque altro?


````
def num_pairwise_keys(n):
    return n * (n - 1) // 2   # C(n, 2): una chiave per ogni coppia non ordinata

for n in [2, 5, 10, 50, 100, 1000]:
    print(f"N = {n:>5} peer  ->  {num_pairwise_keys(n):>10} chiavi pre-condivise necessarie")
````

### Esplorazione interattiva: il grafo delle chiavi

Ogni **arco** nel grafo qui sotto rappresenta una chiave simmetrica pre-condivisa tra due peer. Muovi lo
slider e osserva quanto velocemente il grafo diventa denso — e quanto cresce il numero di chiavi.


````
def draw_peer_key_graph(n=8):
    fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

    # --- Pannello sinistro: grafo dei peer con tutte le chiavi pairwise ---
    ax = axes[0]
    angles = np.linspace(0, 2*np.pi, n, endpoint=False)
    xs, ys = np.cos(angles), np.sin(angles)
    for i, j in itertools.combinations(range(n), 2):
        ax.plot([xs[i], xs[j]], [ys[i], ys[j]], color='tab:blue', alpha=0.3, linewidth=0.8, zorder=1)
    ax.scatter(xs, ys, s=250, color='tab:orange', zorder=2, edgecolor='black')
    for i in range(n):
        label = f"P{i+1}" if n <= 20 else ""
        ax.annotate(label, (xs[i], ys[i]), ha='center', va='center', fontsize=8, zorder=3)
    ax.set_xlim(-1.4, 1.4); ax.set_ylim(-1.4, 1.4); ax.set_aspect('equal'); ax.axis('off')
    n_keys = num_pairwise_keys(n)
    ax.set_title(f"N = {n} peer  ->  {n_keys} chiavi\n(ogni arco = una chiave simmetrica pre-condivisa)")

    # --- Pannello destro: curva di crescita, con il punto corrente evidenziato ---
    ax2 = axes[1]
    n_range = np.arange(2, 61)
    keys_range = n_range * (n_range - 1) // 2
    ax2.plot(n_range, keys_range, color='tab:blue', label="C(N,2) = N(N-1)/2")
    ax2.scatter([n], [n_keys], color='tab:red', zorder=5, s=80, label=f"N={n} -> {n_keys} chiavi")
    ax2.set_xlabel("numero di peer N"); ax2.set_ylabel("numero di chiavi necessarie")
    ax2.set_title("Crescita del numero di chiavi pre-condivise")
    ax2.legend(); ax2.grid(True, alpha=0.3)

    plt.tight_layout(); plt.show()

interact(draw_peer_key_graph, n=IntSlider(value=8, min=2, max=40, step=1, description="N peer"));
````

### Attenzione ai nomi: è un'esplosione *combinatoria*, non *esponenziale*

Il numero di chiavi cresce come $\binom{N}{2} = \frac{N(N-1)}{2}$: questo è **quadratico** in $N$, non
esponenziale. È importante essere precisi con la terminologia parlando con un pubblico tecnico — ma
"quadratico" qui è comunque disastroso in pratica: a 100 peer servono 4.950 chiavi, a 1.000 peer quasi
mezzo milione. E per ogni peer aggiunto, ogni altro peer della rete deve generare, distribuire e
custodire *una nuova chiave* — il problema non è solo il totale, è anche il costo di **manutenzione**
(revoca, rotazione, compromissione di un singolo peer) che ricade su tutti gli altri.


````
# Quanto tocca a un SINGOLO peer, in termini di chiavi da gestire?
n_range = np.arange(2, 1001)
keys_per_peer = n_range - 1   # ogni peer ha una chiave con ciascuno degli altri N-1
total_keys = n_range * (n_range - 1) // 2

fig, ax = plt.subplots(figsize=(6,4))
ax.plot(n_range, total_keys, label="chiavi totali nel sistema (quadratico)")
ax.plot(n_range, keys_per_peer, label="chiavi gestite da UN peer (lineare)")
ax.set_xlabel("numero di peer N"); ax.set_ylabel("numero di chiavi")
ax.set_yscale('log')
ax.set_title("Anche il carico per singolo peer cresce, seppur linearmente")
ax.legend(); ax.grid(True, alpha=0.3)
plt.show()
````

**Takeaway**: la pre-condivisione pairwise funziona per un piccolo gruppo statico (es. due server che
si parlano da sempre), ma non scala a reti aperte o dinamiche (internet, un'azienda con migliaia di
dipendenti, dispositivi IoT che si scoprono a runtime). Serve un modo per **generare** un segreto condiviso
al momento, su un canale che l'attaccante può ascoltare — idealmente usando solo ciò che già abbiamo:
crittografia simmetrica.


## 3. Si può fare key exchange con la sola crittografia simmetrica? Merkle Puzzles

**Idea (Merkle, 1974)**: Alice prepara $N$ "puzzle": ciascuno è una chiave di sessione cifrata con una
chiave-puzzle volutamente debole (pochi bit). Li invia tutti a Bob, in chiaro, sul canale insicuro. Bob ne
sceglie uno a caso e lo rompe per forza bruta (costo $O(K)$, con $K$ = spazio delle chiavi-puzzle). Da quel
momento Alice e Bob condividono la chiave di sessione contenuta in quel puzzle — **senza essersi mai
incontrati prima**, e usando solo un cifrario simmetrico.

**Il punto cruciale**: un attaccante Eve, per trovare la STESSA chiave di sessione che useranno Alice e
Bob, deve nel caso peggiore rompere tutti gli $N$ puzzle: costo $O(N \cdot K)$. Con $K \approx N$
(dimensionamento tipico), il costo di Eve è $O(N^2)$ contro l'$O(N)$ di Bob: un gap **quadratico**.


````
def make_puzzles(n_puzzles, key_bits):
    """Alice genera n_puzzles puzzle. Ognuno cela una chiave di sessione dietro
    una chiave-puzzle debole a key_bits bit, cifrata con AES-ECB (va bene per la demo:
    qui vogliamo solo mostrare il costo di brute force, non la sicurezza di AES)."""
    puzzles, session_keys = [], []
    for i in range(n_puzzles):
        weak_key = random.getrandbits(key_bits).to_bytes(16, 'big')
        session_key = os.urandom(16)
        plaintext = pad(b"PUZZLE#" + i.to_bytes(4, 'big') + session_key, 16)
        ct = AES.new(weak_key, AES.MODE_ECB).encrypt(plaintext)
        puzzles.append(ct)
        session_keys.append(session_key)
    return puzzles, session_keys

def try_open(ct, guess_key_int):
    k = guess_key_int.to_bytes(16, 'big')
    try:
        pt = unpad(AES.new(k, AES.MODE_ECB).decrypt(ct), 16)
    except Exception:
        return None
    return pt if pt.startswith(b"PUZZLE#") else None

def bob_solves(puzzles, key_bits):
    """Bob sceglie UN puzzle a caso e lo rompe per forza bruta: costo atteso O(2^key_bits / 2)."""
    idx = random.randrange(len(puzzles))
    for guess in range(2**key_bits):
        pt = try_open(puzzles[idx], guess)
        if pt is not None:
            return idx, pt[11:], guess + 1
    return None, None, None

def eve_breaks_all(puzzles, key_bits):
    """Eve non sa quale puzzle useranno: nel caso peggiore deve romperli tutti."""
    attempts = 0
    for ct in puzzles:
        for guess in range(2**key_bits):
            attempts += 1
            if try_open(ct, guess) is not None:
                break
    return attempts

KEY_BITS = 14  # piccolo apposta per farlo girare in aula in pochi secondi
N_PUZZLES = 6
puzzles, keys = make_puzzles(N_PUZZLES, KEY_BITS)
idx, session_key, bob_attempts = bob_solves(puzzles, KEY_BITS)
print(f"Bob ha aperto il puzzle #{idx} in {bob_attempts} tentativi.")
print("Chiave di sessione condivisa correttamente (Alice e Bob, mai incontrati prima):",
      session_key == keys[idx])
````

````
t0 = time.time()
eve_attempts = eve_breaks_all(puzzles, KEY_BITS)
t1 = time.time()
print(f"Eve ha dovuto fare {eve_attempts} tentativi in totale ({t1 - t0:.2f}s) "
      f"contro i ~{2**KEY_BITS // 2} attesi di Bob per un solo puzzle.")
````

### Quanto serve alzare i parametri per rendere la vita difficile a Eve?


````
Ks = [10, 12, 14, 16, 18, 20]
bob_cost = [2**k / 2 for k in Ks]              # per puzzle, un solo tentativo di Bob
eve_cost_N10 = [10 * (2**k / 2) for k in Ks]   # con N=10 puzzle, caso peggiore di Eve
eve_cost_N1000 = [1000 * (2**k / 2) for k in Ks]  # con N=1000 puzzle

plt.figure(figsize=(6,4))
plt.plot(Ks, bob_cost, marker='o', label="Costo Bob (1 puzzle)")
plt.plot(Ks, eve_cost_N10, marker='o', label="Costo Eve (N=10 puzzle)")
plt.plot(Ks, eve_cost_N1000, marker='o', label="Costo Eve (N=1000 puzzle)")
plt.yscale('log')
plt.xlabel("bit della chiave-puzzle (K)")
plt.ylabel("tentativi attesi (scala log)")
plt.title("Merkle Puzzles: il gap Bob/Eve è solo un fattore N, non esponenziale")
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()

print(f"Per dare a Eve un costo di ~2^40 con K=20 bit, servono N = {2**40 / (2**20/2):.0e} puzzle:")
print("Alice dovrebbe INVIARE quel numero di puzzle sul canale. Non è praticabile.")
````

**Takeaway di questo file**: abbiamo due strade, entrambe con problemi seri:

1. **Pre-condivisione pairwise**: nessuna crittografia asimmetrica richiesta, ma esplosione combinatoria
   $O(N^2)$ nel numero di chiavi da generare/distribuire/custodire.
2. **Merkle Puzzles**: nessuna pre-condivisione richiesta, ma il gap di sicurezza tra il legittimo
   possessore del segreto ed un attaccante è solo *quadratico* — per essere sicuro, il traffico scambiato
   deve crescere enormemente.

In entrambi i casi, il problema di fondo è lo stesso: **con la sola crittografia simmetrica non esiste un
modo efficiente per instaurare un segreto condiviso tra due parti che non si sono mai incontrate**. Nel
prossimo file vedremo perché serve un oggetto matematico diverso — una **funzione trapdoor** — e come RSA
la realizzi, ottenendo finalmente un gap di sicurezza esponenziale invece che quadratico.
