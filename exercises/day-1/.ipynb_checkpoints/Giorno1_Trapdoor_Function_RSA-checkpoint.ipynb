# Dalla One-Way Function alla Trapdoor Function

## Obiettivo

Partiamo dall'idea vista con Merkle Puzzle:

$$
\text{calcolare} \rightarrow \text{facile}
$$

$$
\text{invertire} \rightarrow \text{difficile}
$$

Ora vogliamo qualcosa di più forte:

$$
\boxed{\text{difficile senza segreto}}
\qquad
\boxed{\text{facile con segreto}}
$$

Questa è l'idea alla base delle **trapdoor function** e ci porterà a RSA.

## 1. One-way function

Consideriamo:

$$
f(x)=x^2
$$

Calcolare $f(x)$ è facile. Ma in questo esempio anche l'inversione è facile.

Cerchiamo quindi una funzione in cui il problema inverso sia molto più difficile.

````python
# Esponenziazione modulare

x = 7
e = 13
n = 55

y = pow(x, e, n)

print(f"x = {x}")
print(f"e = {e}")
print(f"n = {n}")
print(f"y = x^e mod n = {y}")
````

## 2. Aritmetica modulare

$$
a \equiv b \pmod n
$$

quando:

$$
a \bmod n=b \bmod n
$$

Esempi:

````python
examples = [(17, 12), (29, 12), (100, 7), (12345, 97)]

for a, n in examples:
    print(f"{a} mod {n} = {a % n}")
````

### Esponenziazione modulare

Consideriamo:

$$
f(x)=x^e\bmod n
$$

Ad esempio:

$$
y=3^7\bmod55
$$

````python
x = 3
e = 7
n = 55

y = pow(x, e, n)
print("y =", y)
````

## 3. Guardiamo il problema inverso

Conosciamo:

$$
y=x^e\bmod n
$$

e vogliamo trovare $x$.

Con numeri piccoli possiamo fare brute force.

````python
solutions = []

for candidate in range(n):
    if pow(candidate, e, n) == y:
        solutions.append(candidate)

print("y =", y)
print("Possibili preimmagini:", solutions)
````

L'obiettivo crittografico è costruire una situazione in cui:

$$
x\rightarrow f(x)
$$

sia facile, mentre:

$$
y\rightarrow x
$$

sia computazionalmente difficile.

# 4. Trapdoor Function

Una **trapdoor function** aggiunge un'informazione segreta $t$.

$$
y=f(x)
$$

Senza trapdoor:

$$
y\rightarrow x
\qquad\text{difficile}
$$

Con trapdoor:

$$
(y,t)\rightarrow x
\qquad\text{facile}
$$

```text
                         f(x)
              x ─────────────────► y
                                    │
                       ┌────────────┴────────────┐
                       │                         │
                    Eve                       Bob
                senza trapdoor             con trapdoor
                       │                         │
                       ▼                         ▼
                   difficile                    facile
                       │                         │
                       └────────────┬────────────┘
                                    ▼
                                    x
```

La trapdoor è la **scorciatoia segreta** che rende efficiente l'inversione.

# 5. Esempio: moltiplicazione e fattorizzazione

Consideriamo:

$$
p=11,\qquad q=17
$$

Calcolare:

$$
n=pq
$$

è facile.

````python
p = 11
q = 17
n = p * q

print("p =", p)
print("q =", q)
print("n =", n)
````

Il problema inverso è:

$$
187\rightarrow ?\cdot?
$$

Cerchiamo i fattori:

````python
n = 187

factors = []

for p in range(2, n):
    if n % p == 0:
        factors.append((p, n // p))

print("Fattori:", factors)
````

Per numeri piccoli è facile. Per numeri primi sufficientemente grandi, la fattorizzazione è un problema computazionalmente difficile.

L'idea che ci interessa è:

$$
p,q\rightarrow pq \quad \text{facile}
$$

$$
pq\rightarrow p,q \quad \text{difficile}
$$

RSA sfrutta questa asimmetria.

# 6. La struttura segreta

In RSA:

$$
n=pq
$$

è pubblico.

La conoscenza di:

$$
p,q
$$

è invece privata.

Con $p$ e $q$ possiamo calcolare:

$$
\varphi(n)=(p-1)(q-1)
$$

````python
p = 11
q = 17

n = p * q
phi = (p - 1) * (q - 1)

print("n   =", n)
print("phi =", phi)
````

## 7. Esempio RSA completo

Scegliamo:

$$
p=11,\qquad q=17
$$

quindi:

$$
n=187
$$

e:

$$
\varphi(n)=10\cdot16=160
$$

Scegliamo:

$$
e=7
$$

e cerchiamo $d$ tale che:

$$
ed\equiv1\pmod{\varphi(n)}
$$

cioè:

$$
7d\equiv1\pmod{160}
$$

````python
from math import gcd

p = 11
q = 17

n = p * q
phi = (p - 1) * (q - 1)

e = 7
assert gcd(e, phi) == 1

d = pow(e, -1, phi)

print("n   =", n)
print("phi =", phi)
print("e   =", e)
print("d   =", d)
print("e*d mod phi =", (e * d) % phi)
````

Abbiamo:

### Chiave pubblica

$$
(n,e)=(187,7)
$$

### Chiave privata

$$
(n,d)=(187,23)
$$

perché:

$$
7\cdot23\equiv1\pmod{160}
$$

# 8. RSA: cifratura

Cifriamo:

$$
m=42
$$

usando:

$$
c=m^e\bmod n
$$

````python
m = 42

c = pow(m, e, n)

print("Messaggio :", m)
print("Cifrato   :", c)
````

# 9. RSA: decifratura

Usiamo la chiave privata:

$$
m=c^d\bmod n
$$

````python
m_decrypted = pow(c, d, n)

print("Cifrato   :", c)
print("Decifrato :", m_decrypted)

assert m_decrypted == m
````

## 10. La trapdoor in RSA

```text
                     CHIAVE PUBBLICA
                         (n, e)
                            │
                            ▼
        m ───────────► c = m^e mod n
                            │
                            ▼
                            c
                            │
              ┌─────────────┴─────────────┐
              │                           │
            Eve                          Bob
       solo (n,e)                    conosce d
              │                           │
              ▼                           ▼
       inversione                   m = c^d mod n
         difficile                      facile
```

La differenza fondamentale rispetto a Merkle è:

$$
\boxed{\text{Bob possiede una scorciatoia segreta}}
$$

Questa è l'idea della **trapdoor**.

# Takeaway

### One-way

$$
x\rightarrow f(x) \quad \text{facile}
$$

$$
f(x)\rightarrow x \quad \text{difficile}
$$

### Trapdoor

$$
x\rightarrow f(x) \quad \text{facile}
$$

$$
f(x)\rightarrow x \quad \text{difficile senza }t
$$

$$
(f(x),t)\rightarrow x \quad \text{facile}
$$

### RSA

La struttura:

$$
n=pq
$$

permette di costruire una funzione con questa proprietà.

**Prossimo passo:** capire $φ(n)$, l'inverso modulare e perché funziona la decifratura RSA.
