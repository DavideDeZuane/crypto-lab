# Logaritmo Discreto

Alcuni richiami sul problema del logaritmo discreto.  

- Lavoriamo in $(\mathbb{Z}^*_p, *)$, con $p$ primo di $n$ bits
  
- Ci serve una **radice primitiva** $g$ modulo $p$, ovvero $$g \in \mathbb{Z}_p^* \quad \text{t.c.} \quad \mathbb{Z}_p^* = \{g^0, g^1, g^2, \ldots, g^{p-1} \}$$


**In tale scenario**:

$$
g^x \quad \text{facile da calcolare}
$$

$$
x \leftarrow g^x \quad \text{difficile da recuperare}
$$

# Diffie-Hellman Exchange

Piccoli richiami sullo scambio Diffie-Hellman

- $g$ e $q$, sono parametri di sistema che sono noti a tutte le parti coinvolte nella comunicazione.  
   
- Alice sceglie un esponente segreto $\alpha \xleftarrow{R} \mathbb{Z}_p$, calcola $u \leftarrow g^\alpha$ e lo manda a Bob.  

- Bob sceglie un esponente segreto $\beta \xleftarrow{R} \mathbb{Z}_p$, calcola $v \leftarrow g^\beta$ e lo manda a Alice.  

- Entrambe le parti calcolano rispettivamente $v^\alpha$ e $u^\beta$, e il segreto condiviso è $w=v^\alpha = g^{\alpha\beta} = u^\beta$.
