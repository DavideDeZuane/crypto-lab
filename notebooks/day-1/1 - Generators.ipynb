
## Radici Primitive


Un elemento $g \in \mathbb{Z}_n^*$ si dice **radice primitiva modulo $n$** del gruppo se, calcolando le sue potenze modulo $n$, possiamo ottenere tutti gli elementi del gruppo.

$$
\langle g \rangle =
\{g^0,g^1,g^2,\ldots\}
=
\mathbb{Z}_n^*
$$


Esempio: considerando $\mathbb{Z}^*_5$, trovare se esistono le radici primitive con brute forcing:

|---:|---:|---:|---:|---:|---:|---:|
| $1$ | $1$ | $1$ | $1$ | $1$ | $1$ | ? |
| $2$ | $1$ | $2$ | $4$ | $3$ | $1$ | ? |
| $3$ | $1$ | $3$ | $4$ | $2$ | $1$ | ? |



## Possiamo fare meglio?

Possiamo usare dei **safe-prime** che si basano sullo scegliere $ p = 2q+1 $ dove anche $q$ è primo.   
E ci basta verificare queste due condizioni:

$$ g^{(p-1)/2} \not\equiv 1 \pmod p, \quad g^{(p-1)/q} \not\equiv 1 \pmod p$$

- Confrontiamo questo metodo con quello a brute force

```python
# Confronto
test_primes = [23, 101, 503, 727, 1231, 2003, 3463]
compare_primitive_roots(test_primes, True)
```
