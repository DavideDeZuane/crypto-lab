# Digital Signature Algorithm 


A signature scheme $S = (G, S, V )$ is a triple of efficient algorithms, $G$, $S$ and $V$ , where 

- $G$ is called a key *generation algorithm*,
- $S$ is called a *signing algorithm*,
- $V$ is called a *verification algorithm*.

Algorithm $S$ is used to generate signatures and algorithm $V$ is used to verify signatures.

## RSA Signature

**Keygen**

- Alice genera due primi $p$, $q$ e calcola $n = pq$
  
- Sceglie esponente $e$

- Calcola esponente inverso $d$

**Sign**

- Dato $m$, messaggio non confidenziale da firmare

- Alice firma calcolando: $\ y = m^d \mod n$

- Pubblica la coppia $(m,\ y)$

**Verify**

- Bob vuole verificare la firma

- Quindi calcola  $\ m^` = y^e \mod n$

- E verifica che $\ m = m^`$

## El-Gamal Signature 


**Keygen**

- Alice sceglie $p$, una radice primitiva $\alpha$
- Sceglie numero segreto $x \in [1, p-2]$
- Calcola $\beta = \alpha^x \mod p$

**Sign**

- Dato $m$ messaggio da firmare
- Alice sceglie un numero $k$ tale che $\ gcd(k,p-1) = 1$
- Calcola $r = \alpha^k \mod p$ e $s = k^{-1}(m - x\cdot r) \mod p-1$, dove $k\cdot k^{-1} = 1 \mod p-1$
- La firma è composta :
$$(m,\ r,\ s)$$

**Verify**

- Per verificare la firma
- Bob calcola $$v_1 = \beta^r r^s \mod p,\quad v_2 = \alpha^m \mod p$$
- La firma per Bob è valido se e solo se $v_1 = v_2 \mod p$
