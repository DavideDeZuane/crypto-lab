```python
import sys
import mpmath as mp

# Options for large parameters
sys.set_int_max_str_digits(10000)
mp.mp.dps = 2600 

toy_parameters = {
    "TOY-16": {
        "bits": 17,
        "p": 65537,
        "g": 3,
    },

    "TOY-32": {
        "bits": 32,
        "p": 4294967291,
        "g": 2,
    },
}


# Verificare se g e' un generatore del gruppo



# Come genero gli esponenti segreti






# Parametri reali 


def generate_modp_prime(bits, c):
    """
    Generate the RFC 3526 MODP prime:

    p = 2^bits - 2^(bits-64) - 1 + 2^64 * floor(2^(bits-130) * pi) + 2^64 * c
    """

    k = bits - 130

    # High-precision pi
    pi = mp.pi

    # floor(2^k * pi)
    floor_pi = int(mp.floor(mp.power(2, k) * pi))

    p = (
        2**bits
        - 2**(bits - 64)
        - 1
        + 2**64 * (floor_pi + c)
    )

    return p



modp_parameters = {
    "MODP-1536": {
        "bits": 1536,
        "p": generate_modp_prime(
            1536,
            741804
        ),
        "g": 2,
        "rfc_group": 5,
    },

    "MODP-2048": {
        "bits": 2048,
        "p": generate_modp_prime(
            2048,
            124476
        ),
        "g": 2,
        "rfc_group": 14,
    },

    "MODP-3072": {
        "bits": 3072,
        "p": generate_modp_prime(
            3072,
            1690314
        ),
        "g": 2,
        "rfc_group": 15,
    },

    "MODP-4096": {
        "bits": 4096,
        "p": generate_modp_prime(
            4096,
            240904
        ),
        "g": 2,
        "rfc_group": 16,
    },

    "MODP-6144": {
        "bits": 6144,
        "p": generate_modp_prime(
            6144,
            929484
        ),
        "g": 2,
        "rfc_group": 17,
    },

    "MODP-8192": {
        "bits": 8192,
        "p": generate_modp_prime(
            8192,
            4743158
        ),
        "g": 2,
        "rfc_group": 18,
    },
}


# Configura qui il profilo che desideri testare
GROUP = "MODP-2048"

# Estrazione sicura dei dati
if GROUP in modp_parameters:
    gruppo = modp_parameters[GROUP]
    p = gruppo["p"]
    g = gruppo["g"]
    dimensione_bit = gruppo["bits"]
    
    print(f"[Profilo Caricato]: {GROUP} ({dimensione_bit} bit)")
    print(f"[Public Values]\np = {p}\ng = {g}")
else:
    print(f"Errore: Il profilo '{GROUP}' non è presente nella tabella.")

x = 7
A = pow(g, x, p)
print(f"[ALICE] Secret Key, x = {x}")
print(f"[ALICE] Public Value, g^x = {A} ")
print()
print(f"[ALICE] ---- g^x ----> BOB")
print()
y = 15
B = pow(g, y, p)
print(f"[BOB] Secret Key, y = {y}")
print(f"[BOB] Public Value, g^y = {B} ")
print()
print(f"[ALICE] <--- g^y ----- BOB")
print()
# Both compute the shared secret
K_Alice = pow(B, x, p)
K_Bob = pow(A, y, p)
print(f"[ALICE] Shared secret, (g^y)^x = {K_Alice} ")
print(f"[ALICE] Shared secret, (g^x)^y = {K_Bob} ")

print("\nSame secret?", K_Alice == K_Bob)
```
