```python


toy_parameters = {

    "TOY-10": {
        "bits": 10,
        "p": 1009,
        "g": 11,
    },

    "TOY-14": {
        "bits": 14,
        "p": 12289,
        "g": 11,
    },
    
    "TOY-16": {
        "bits": 17,
        "p": 65537,
        "g": 3,
    },

    "TOY-24": {
        "p": 16777259,
        "g": 2,
    },

    "TOY-32": {
        "bits": 32,
        "p": 4294967291,
        "g": 2,
    },
}


# Verificare se g e' un generatore del gruppo



# Come genero gli esponenti segreti



GROUP = "TOY-16"

# Estrazione sicura dei dati
if GROUP in toy_parameters:
    gruppo = toy_parameters[GROUP]
    p = gruppo["p"]
    g = gruppo["g"]
    dimensione_bit = gruppo["bits"]
    
    print(f"[Profilo Caricato]: {GROUP} ({dimensione_bit} bit)")
    print(f"[Public Values]\np = {p}\ng = {g}")
else:
    print(f"Errore: Il profilo '{GROUP}' non è presente nella tabella.")

x = 7
A = pow(g, x, p)
print(f"[ALICE] Secret Key, x = {x}")
print(f"[ALICE] Public Value, g^x = {A} ")
print()
print(f"[ALICE] ---- g^x ----> BOB")
print()
y = 15
B = pow(g, y, p)
print(f"[BOB] Secret Key, y = {y}")
print(f"[BOB] Public Value, g^y = {B} ")
print()
print(f"[ALICE] <--- g^y ----- BOB")
print()
# Both compute the shared secret
K_Alice = pow(B, x, p)
K_Bob = pow(A, y, p)
print(f"[ALICE] Shared secret, (g^y)^x = {K_Alice} ")
print(f"[ALICE] Shared secret, (g^x)^y = {K_Bob} ")

print("\nSame secret?", K_Alice == K_Bob)
```

```python
import time

def discrete_log_bruteforce(g, A, p):
    for x in range(p):
        if pow(g, x, p) == A:
            return x

    return None

p = 16777259
g = 2
# scegliamo un esponente abbastanza grande almeno aumentiamo il numero di try
a = p // 2

A = pow(g, a, p)

start = time.perf_counter()

recovered = discrete_log_bruteforce(g, A, p)

elapsed = time.perf_counter() - start

print(f"Recovered: {recovered}")
print(f"Time: {elapsed:.2f} seconds")

# A questo punto non mi serve neanche rompere quello di bob ma mi basta moltiplicarlo con l'esponente segreto
```
