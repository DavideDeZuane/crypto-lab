```python
# ============================================
# CURVA ELLITTICA - APPROCCIO SCHOOLBOOK
#
# E: y^2 = x^3 + ax + b (mod p)
# ============================================


# --------------------------------------------
# 1. INVERSO MODULARE
# --------------------------------------------

def inverse_mod(x, p):
    """
    Restituisce l'inverso di x modulo p.

    Trova y tale che:
        x * y = 1 (mod p)
    """
    return pow(x, -1, p)


# --------------------------------------------
# 2. ADDIZIONE DI PUNTI
# --------------------------------------------

def point_add(P, Q, a, p):
    """
    Calcola P + Q sulla curva:

        y^2 = x^3 + ax + b (mod p)

    Il punto all'infinito O è rappresentato da None.
    """

    # O + Q = Q
    if P is None:
        return Q

    # P + O = P
    if Q is None:
        return P

    x1, y1 = P
    x2, y2 = Q

    # ----------------------------------------
    # P + (-P) = O
    # ----------------------------------------

    if x1 == x2 and (y1 + y2) % p == 0:
        return None

    # ----------------------------------------
    # Caso P != Q
    # ----------------------------------------

    if P != Q:

        # lambda = (y2-y1)/(x2-x1)

        numerator = y2 - y1
        denominator = x2 - x1

        denominator_inv = inverse_mod(
            denominator % p,
            p
        )

        lam = (
            numerator * denominator_inv
        ) % p

    # ----------------------------------------
    # Caso P == Q
    # Raddoppiamento: P + P
    # ----------------------------------------

    else:

        # lambda = (3*x1^2 + a)/(2*y1)

        numerator = 3 * x1**2 + a
        denominator = 2 * y1

        # Se il denominatore è 0,
        # la tangente è verticale e:
        #
        # P + P = O

        if denominator % p == 0:
            return None

        denominator_inv = inverse_mod(
            denominator % p,
            p
        )

        lam = (
            numerator * denominator_inv
        ) % p

    # ----------------------------------------
    # Calcolo di x3
    # ----------------------------------------

    x3 = (
        lam**2
        - x1
        - x2
    ) % p

    # ----------------------------------------
    # Calcolo di y3
    # ----------------------------------------

    y3 = (
        lam * (x1 - x3)
        - y1
    ) % p

    return (x3, y3)


# --------------------------------------------
# 3. TROVA TUTTI I PUNTI
# --------------------------------------------

def find_all_points(a, b, p):
    """
    Cerca tutti i punti della curva provando
    TUTTE le possibili coppie (x, y).

    Questo è volutamente un approccio
    brute-force / schoolbook.
    """

    points = []

    for x in range(p):

        for y in range(p):

            left = (y * y) % p

            right = (
                x**3
                + a * x
                + b
            ) % p

            # Controlliamo:
            #
            # y^2 = x^3 + ax + b (mod p)

            if left == right:
                points.append((x, y))

    return points


# --------------------------------------------
# 4. ORDINE DI UN PUNTO
# --------------------------------------------

def point_order(P, a, p):
    """
    Calcola l'ordine di P.

    L'ordine è il più piccolo intero n > 0
    tale che:

        nP = O
    """

    # L'ordine di O è 1
    if P is None:
        return 1

    Q = P
    order = 1

    while True:

        Q = point_add(Q, P, a, p)

        order += 1

        if Q is None:
            return order


# --------------------------------------------
# 5. TROVA I GENERATORI
# --------------------------------------------

def find_generators(points, a, p):
    """
    Trova tutti i punti che generano l'intero gruppo.

    Un punto P è generatore se:

        ordine(P) = |E|
    """

    group_order = len(points)

    generators = []

    for P in points:

        # O non può essere un generatore
        if P is None:
            continue

        order = point_order(P, a, p)

        if order == group_order:
            generators.append(P)

    return generators


# ============================================
# MAIN
# ============================================

if __name__ == "__main__":

    # ----------------------------------------
    # PARAMETRI DELLA CURVA
    # ----------------------------------------

    a = 2
    b = 2
    p = 17

    print("Curva:")
    print(
        f"y^2 = x^3 + {a}x + {b} (mod {p})"
    )

    print()

    # ----------------------------------------
    # TROVIAMO TUTTI I PUNTI
    # ----------------------------------------

    points = find_all_points(a, b, p)

    # Aggiungiamo il punto all'infinito
    points.append(None)

    print("Punti della curva:")
    print()

    for P in points:

        if P is None:
            print("O")

        else:
            print(P)

    print()
    print(
        "Numero di punti:",
        len(points)
    )

    # ----------------------------------------
    # CALCOLIAMO L'ORDINE DI OGNI PUNTO
    # ----------------------------------------

    print()
    print("Ordine dei punti:")
    print()

    for P in points:

        if P is None:
            continue

        order = point_order(
            P,
            a,
            p
        )

        print(
            f"{P} -> ordine {order}"
        )

    # ----------------------------------------
    # CERCHIAMO I GENERATORI
    # ----------------------------------------

    generators = find_generators(
        points,
        a,
        p
    )

    print()
    print("Generatori:")
    print()

    for P in generators:
        print(P)
```
